**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python
# Inizializza il client di Open AI
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')  # Carica le variabili dal file .env

client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'))
```

```python
input_text = input("Enter text (premi invio per articolo di notizie di esempio): ")
if not input_text.strip():
    input_text = (
        "L'Agenzia Spaziale Europea ha annunciato oggi con successo il lancio della nuova missione diretta verso Giove. "
        "La sonda spaziale studierà approfonditamente le lune ghiacciate per individuare la presenza di oceani d'acqua liquida sotterranei. "
        "Il viaggio richiederà circa otto anni di navigazione nello spazio profondo e vede la collaborazione scientifica di trenta paesi."
    )

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente editoriale specializzato nella sintesi giornalistica. "
                "Riassumi l'articolo fornito in un singolo paragrafo che colga gli elementi essenziali. "
                "VINCOLO TASSATIVO: la lunghezza complessiva del testo generato deve essere rigorosamente inferiore o uguale a 255 caratteri (inclusi gli spazi). "
                "Restituisci esclusivamente il riassunto, senza preamboli, commenti o titoli."
            ),
        },
        {
            "role": "user",
            "content": input_text,
        },
    ],
)

summary = complention.choices[0].message.content.strip()
print(summary)
print(f"\nConteggio caratteri: {len(summary)}/255")
```
