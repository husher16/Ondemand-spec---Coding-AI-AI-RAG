**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese con `ollama.chat`.

```python
# Importa la libreria Ollama
import ollama
```

```python
input_text = input("Enter text (premi invio per descrizione prodotto di esempio): ")
if not input_text.strip():
    input_text = "Questo smartwatch impermeabile con display AMOLED da 1.4 pollici offre monitoraggio del sonno, frequenza cardiaca e un'autonomia della batteria fino a 10 giorni."

response = ollama.chat(
    model="llama3.2",  # Modifica con il modello installato localmente in Ollama (es. llama3.2, mistral, gemma2)
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un traduttore professionista bilingue specializzato in traduzioni tra italiano e inglese. "
                "Se il testo fornito è in italiano, traducilo in un inglese naturale e fluido. "
                "Se il testo è in inglese, traducilo in un italiano corretto e naturale. "
                "Mantieni il tono promozionale e chiaro adatto a descrizioni di prodotto. "
                "Restituisci esclusivamente la traduzione, senza note o spiegazioni aggiuntive."
            ),
        },
        {
            "role": "user",
            "content": input_text,
        },
    ],
)

print(response["message"]["content"])
```
