**Traccia:** Crea un prompt che estragga i dettagli chiave (nome, età, professione) da un testo descrittivo su una persona.

**Obiettivo:** Estrarre informazioni strutturate da un testo libero.

```python
# Inizializza il client di Open AI
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')  # Carica le variabili dal file .env

client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'))
```

```python
input_text = input("Enter text (premi invio per testo descrittivo di esempio): ")
if not input_text.strip():
    input_text = "Mi chiamo Sofia Martini, ho 28 anni e lavoro come ingegnere aerospaziale presso un centro di ricerca a Torino."

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente esperto nell'estrazione di informazioni strutturate da testi liberi. "
                "Estrai con precisione dal testo i seguenti dettagli chiave relativi alla persona: "
                "Nome, Età, Professione. "
                "Formatta la risposta in JSON strutturato con le chiavi 'nome', 'eta' e 'professione'. "
                "Se un'informazione non è menzionata nel testo, assegna il valore 'Non specificato'."
            ),
        },
        {
            "role": "user",
            "content": input_text,
        },
    ],
)

print(complention.choices[0].message.content)
```
