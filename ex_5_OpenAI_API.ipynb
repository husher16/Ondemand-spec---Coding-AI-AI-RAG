**Obiettivo:** Sviluppare un prompt per creare domande educative basate su un contenuto.

```python
# Inizializza il client di Open AI
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')  # Carica le variabili dal file .env

client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'))
```

```python
input_text = input("Enter text (premi invio per testo di esempio): ")
if not input_text.strip():
    input_text = (
        "La fotosintesi clorofilliana è il processo biologico mediante il quale le piante verdi, le alghe "
        "e alcuni batteri trasformano l'energia solare in energia chimica. Catturando acqua dal terreno e anidride "
        "carbonica dall'atmosfera, sintetizzano molecole di glucosio per il proprio nutrimento, liberando ossigeno come sottoprodotto."
    )

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un docente specializzato nella formulazione di verifiche educative e quesiti di comprensione del testo. "
                "Dato il testo fornito, genera una lista di 3-4 domande educative di comprensione per valutare la conoscenza "
                "dei concetti cardine. Per ogni domanda, fornisci chiaramente sia il quesito sia la risposta corretta "
                "estratta fedelmente dalle informazioni del testo."
            ),
        },
        {
            "role": "user",
            "content": f"Testo di riferimento:\n\n{input_text}",
        },
    ],
)

print(complention.choices[0].message.content)
```
