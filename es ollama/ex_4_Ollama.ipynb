**Traccia:** Trasforma un set di dati strutturati (come un dizionario Python) in una descrizione testuale completa.

**Obiettivo:** Convertire dati strutturati in una descrizione leggibile con `ollama.chat`.

```python
# Importa la libreria Ollama
import ollama
```

```python
input_text = input("Enter text (oppure premi invio per usare il dizionario 'data'): ")
data = {"name": "Giovanni", "age": 30, "city": "Roma", "profession": "Ingegnere"}

prompt_content = input_text.strip() if input_text.strip() else f"{data}"

response = ollama.chat(
    model="llama3.2",  # Modifica con il modello installato localmente in Ollama (es. llama3.2, mistral, gemma2)
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente specializzato nella conversione di dati strutturati (es. dizionari, tabelle) "
                "in descrizioni testuali complete, scorrevoli e naturali in italiano. "
                "Integra tutte le informazioni fornite (nome, età, città, professione, ecc.) in modo fluido e coerente, "
                "senza omettere alcun dato e senza aggiungere dettagli non presenti."
            ),
        },
        {
            "role": "user",
            "content": f"Converti questi dati strutturati in una descrizione testuale completa:\n{prompt_content}",
        },
    ],
)

print(response["message"]["content"])
```
